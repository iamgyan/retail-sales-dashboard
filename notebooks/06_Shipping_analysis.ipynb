# 🚚 Shipping Analysis

## Business Objective

Shipping efficiency plays a vital role in customer satisfaction and operational performance.

This analysis aims to answer:

### Business Questions

1. 🚚 Which shipping mode is most frequently used?
2. 💰 Which shipping mode generates the highest revenue?
3. 📈 Which shipping mode is the most profitable?
4. ⏳ Does longer shipping reduce profitability?
5. 🎯 How can shipping performance be improved?


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")

plt.rcParams["figure.figsize"] = (12,6)
df = pd.read_csv('../data/Superstore.csv', encoding="latin1")
```

```python
shipping_summary=(
    df.groupby("Ship Mode").agg(

        Revenue=("Sales","sum"),

        Profit=("Profit","sum"),
    
        Orders=("Order ID","nunique"),
    
        Average_Shipping_Days=("Shipping Days","mean")).sort_values("Revenue",

    ascending=False))

shipping_summary
```
